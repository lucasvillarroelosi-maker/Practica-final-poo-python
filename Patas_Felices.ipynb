```python
# Inventario
class inventario:
  def __init__(self):
    self.lista_productos = []

  def agregar_producto(self, producto):
    self.lista_productos.append(producto)

  def actualizar_inventario(self, producto, cantidad):
    for prod in self.lista_productos:
      if prod.nombre == producto.nombre:
        prod.actualizar_cantidad(cantidad)

  def generar_alerta(self, umbral_minimo):
    alertas = [prod.nombre for prod in self.lista_productos if prod.cantidad <= umbral_minimo]
    return f"Productos por debajo del umbral: {','.join(alertas)}" if alertas else "No hay productos por debajo del umbral"

  def mostrar_productos(self):
    return [producto.mostrar_informacion() for producto in self.lista_productos]
```

```python
from datetime import datetime

# Ventas
class ventas:
  def __init__(self, cliente, lista_productos):
    self.cliente = cliente
    self.lista_productos = lista_productos
    self.fecha = datetime.now()
    self.total = self.calcular_total()

  def calcular_total(self):
    return sum(producto.precio for producto in self.lista_productos)

  def registrar_venta(self):
    self.cliente.registrar_compra(self)
    return f'Venta registrada con exito. Total: {self.mostrar_informacion()}'

  def mostrar_informacion(self):
    productos_str = ", ".join([producto.nombre for producto in self.lista_productos])
    return f"Cliente: {self.cliente.nombre}, Fecha: {self.fecha}, Productos: {productos_str}, Total: {self.total}"
```

```python
# Cliente
class cliente:
  def __init__(self, nombre, direccion, telefono):
    self.nombre = nombre
    self.direccion = direccion
    self.telefono = telefono
    self.historial_compras = []

  def actualizar_informacion(self, direccion=None, telefono=None):
    if direccion:
      self.direccion = direccion
    if telefono:
      self.telefono = telefono

  def registrar_compra(self, compra):
    self.historial_compras.append(compra)

  def mostrar_informacion(self):
    return f'Nombre: {self.nombre}, Direccion: {self.direccion}, Telefono: {self.telefono}'
```

```python
#Productos
class producto:
  def __init__(self, nombre, categoria, precio, cantidad):
    self.nombre = nombre
    self.categoria = categoria
    self.precio = precio
    self.cantidad = cantidad

  def actualizar_cantidad(self, cantidad):
    self.cantidad = cantidad

  def actualizar_precio(self, precio):
    self.precio = precio

  def mostrar_informacion(self):
    return f'Nombre: {self.nombre}, Categoria: {self.categoria}, Cantidad: {self.cantidad}, Precio: {self.precio}'
```

```python
# Mascota
class mascota:
  def __init__(self,nombre, edad, salud, precio):
    self.nombre = nombre
    self.edad = edad
    self.salud = salud
    self.precio = precio

  def actualizar_informacion(self, edad= None, salud=None, precio=None):
    if edad:
      self.edad = edad
    if salud:
      self.salud = salud
    if precio:
      self.precio = precio

  def mostrar_informacion(self):
    return f'Nombre: {self.nombre}, Edad: {self.edad}, Salud: {self.salud}, Precio: {self.precio}'

class perro(mascota):
  def __init__(self, nombre, edad, salud, precio, raza, nivel_de_energia):
    super().__init__(nombre, edad, salud, precio)
    self.raza = raza
    self.nivel_de_energia = nivel_de_energia

  def mostrar_caracteristicas(self):
    return f'Raza: {self.raza}, Nivel de energia: {self.nivel_de_energia}'

class gato(mascota):
  def __init__(self, nombre, edad, salud, precio, raza, independencia):
    super().__init__(nombre, edad, salud, precio)
    self.raza = raza
    self.independencia = independencia

  def mostrar_caracteristicas(self):
    return f'Raza: {self.raza}, Independencia: {self.independencia}'
```

```python
# Main
# Funciones para la interfaz de consola:
def registrar_mascota():
  tipo = input("Ingrese el tipo de mascota (perro/gato): ").strip().lower()
  nombre = input("Ingrese el nombre de la mascota: ")
  edad = int(input("Ingrese la edad de la mascota: "))
  salud = input("Estado de la salud de la mascota: ")
  precio = float(input("Precio de atencion de la mascota: "))

  if tipo == "perro":
     raza = input("Raza del perro: ")
     nivel_de_energia = input("Nivel de energia del perro: ")
     mascota_obj = perro(nombre, edad, salud, precio, raza, nivel_de_energia)
     return mascota_obj
  elif tipo == "gato":
    raza = input("Raza del gato: ")
    independencia = input("Nivel de independencia del gato: ")
    mascota_obj = gato(nombre, edad, salud, precio, raza, independencia)
    return mascota_obj
  else:
    print("Tipo de mascota no reconocida")
    return None

def registrar_producto():
  nombre = input("Nombre del producto: ")
  categoria = input("Categoria del producto: ")
  precio = float(input("Precio del producto: "))
  cantidad = int(input("Cantidad del producto: "))
  producto_obj = producto(nombre, categoria, precio, cantidad)
  return producto_obj

def registrar_cliente():
  nombre = input("Nombre del cliente: ")
  direccion = input("Direccion del cliente: ")
  telefono = input("Telefono del cliente: ")
  cliente_obj = cliente(nombre, direccion, telefono)
  return cliente_obj

def registrar_venta(clientes, inventario):
  nombre_cliente = input("Nombre del cliente: ")
  cliente_encontrado = next((c for c in clientes if c.nombre == nombre_cliente), None)
  if not cliente_encontrado:
    print("Cliente no encontrado")
    return

  productos_seleccionados = []
  while True:
    nombre_producto = input("Nombre del producto (deje vacio para finalizar): ")
    if not nombre_producto:
      break
    prod = next((p for p in inventario.lista_productos if p.nombre == nombre_producto), None)
    if prod:
      productos_seleccionados.append(prod)
      print(f"Producto '{nombre_producto}' agregado a la venta.")
    else:
      print("Producto no encontrado en el inventario")

  if productos_seleccionados:
    venta_obj = ventas(cliente_encontrado, productos_seleccionados)
    print(venta_obj.registrar_venta())
  else:
    print("No se ha registrado la venta porque no seleccionó productos.")

def mostrar_menu():
  print("\n-- Menu de gestion de Patas Felices----")
  print("1. Registrar mascota")
  print("2. Registrar cliente")
  print("3. Registrar producto")
  print("4. Registrar venta")
  print("5. Mostrar informacion acerca de Mascotas")
  print("6. Mostrar informacion acerca de Cliente")
  print("7. Mostrar informacion acerca de Productos")
  print("8. Generar alerta de inventario")
  print("9. Salir")

def main():
  mascotas = []
  clientes = []
  inv = inventario()

  while True:
    mostrar_menu()
    opcion = input("Seleccione una opcion: ").strip()

    if opcion == "1":
      nueva_mascota = registrar_mascota()
      if nueva_mascota:
        mascotas.append(nueva_mascota)
        print("Mascota registrada con exito")
      else:
        print("No se ha registrado la mascota")

    elif opcion == "2":
      nuevo_cliente = registrar_cliente()
      if nuevo_cliente:
        clientes.append(nuevo_cliente)
        print("Cliente registrado con exito")
      else:
        print("No se ha registrado el cliente")

    elif opcion == "3":
      nuevo_producto = registrar_producto()
      if nuevo_producto:
        inv.agregar_producto(nuevo_producto)
        print("Producto registrado con exito")
      else:
        print("No se ha registrado el producto")

    elif opcion == "4":
      registrar_venta(clientes, inv)

    elif opcion == "5":
      print("\n--- Informacion de las mascotas ---")
      for m in mascotas:
        print(m.mostrar_informacion())
        if isinstance(m, perro) or isinstance(m, gato):
          print(m.mostrar_caracteristicas())

    elif opcion == "6":
      print("\n--- Informacion de los clientes ---")
      for c in clientes:
        print(c.mostrar_informacion())

    elif opcion == "7":
      print("\n--- Informacion de los productos ---")
      for p in inv.lista_productos:
        print(p.mostrar_informacion())

    elif opcion == "8":
      umbral_minimo = int(input("Ingrese el umbral minimo de inventario: "))
      print(inv.generar_alerta(umbral_minimo))

    elif opcion == "9":
      print("Saliendo del programa")
      break
    else:
      print("Opcion no valida. Intente de nuevo")
if __name__ == "__main__":
  main()
```
